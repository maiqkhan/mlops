### Baseline model for batch monitoring example

```python
import requests 
import datetime
import pandas as pd
from tqdm import tqdm 
from joblib import load, dump

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error


from evidently import Dataset, DataDefinition, Report, Regression
from evidently.metrics import ValueDrift, DatasetMissingValueCount
from evidently.presets import DataDriftPreset
```

```python
files = [("green_tripdata_2022-02.parquet", "./data"), ("green_tripdata_2022-01.parquet", "./data")]

print("Download files:")
for file, path in files:
    url=f"https://d37ci6vzurychx.cloudfront.net/trip-data/{file}"
    resp=requests.get(url, stream=True)
    save_path=f"{path}/{file}"
    with open(save_path, "wb") as handle:
        for data in tqdm(
                resp.iter_content(),
                desc=f"{file}",
                postfix=f"save to {save_path}",
                total=int(resp.headers["Content-length"])
        ):
            handle.write(data)
           
```

```python
jan_data = pd.read_parquet("data/green_tripdata_2022-01.parquet")
```

```python
jan_data.describe()
```

```python
jan_data.shape
```

```python
# create target
jan_data['duration_min'] = jan_data.lpep_dropoff_datetime - jan_data.lpep_pickup_datetime
jan_data['duration_min'] = jan_data['duration_min'].apply(lambda x: float(x.total_seconds())/ 60)
```

```python
# filter out outliars
jan_data_clean = jan_data.query('duration_min >= 0 and duration_min <= 60 and passenger_count > 0 and passenger_count <= 8 and passenger_count.notnull()', engine='python')
```

```python
jan_data_clean['duration_min'].hist()
```

### Feature selection

```python
target = "duration_min"
num_features = ["passenger_count", "trip_distance", "fare_amount", "total_amount"]
cat_features = ["PULocationID", "DOLocationID"]
```

```python
jan_data.shape
```

```python
train_data = jan_data_clean[:30_000].copy()
val_data = jan_data_clean[30_000:].copy()
```

```python
model = LinearRegression()
```

```python
model.fit(train_data[num_features + cat_features], train_data[target])
```

```python
train_preds = model.predict(train_data[num_features + cat_features])
train_data['prediction'] = train_preds
```

```python
val_preds = model.predict(val_data[num_features + cat_features])
val_data['prediction'] = val_preds
```

```python
print(mean_absolute_error(train_data.duration_min, train_data.prediction))
print(mean_absolute_error(val_data.duration_min, val_data.prediction))
```

#### Evidently Report

```python
data_definition = DataDefinition(
    numerical_columns=num_features + ["prediction"],
    categorical_columns=cat_features,
    regression=[Regression(target=None, prediction="prediction")],
    
    )
```

```python
report = Report(
    metrics=[
        ValueDrift(column="prediction"),
        DataDriftPreset(),
        DatasetMissingValueCount()
    ]
)
```

```python
reference_data = Dataset.from_pandas(train_data, data_definition=data_definition)
current_data = Dataset.from_pandas(val_data, data_definition=data_definition)
```

```python
taxi_eval = report.run(reference_data=reference_data, current_data=current_data)
```

```python
def get_drift_scores_v077(report):
    """Extract drift scores from Evidently 0.7.7 report"""
    
    # Access the metrics results
    results = {}
    
    # Get prediction column drift (ValueDrift metric)
    for metric in report.metrics:
        metric_name = type(metric).__name__
        
        if metric_name == "ValueDrift":
            print(metric.__dir__())
            # Access the drift result for the prediction column
            drift_score = metric.get_result().drift_score
            drift_detected = metric.get_result().drift_detected
            p_value = getattr(metric.get_result(), 'p_value', None)
            
            results['prediction_drift_score'] = drift_score
            results['prediction_drift_detected'] = drift_detected
            results['prediction_p_value'] = p_value
            
            print(f"Prediction column drift score: {drift_score}")
            print(f"Prediction column drift detected: {drift_detected}")
            if p_value is not None:
                print(f"Prediction column p-value: {p_value}")
        
        elif metric_name == "DataDriftPreset":
            # This contains multiple drift metrics for all columns
            dataset_drift = metric.get_result()
            
            # Overall dataset drift information
            results['dataset_drift_score'] = dataset_drift.drift_score
            results['number_of_drifted_columns'] = dataset_drift.number_of_drifted_columns
            results['share_of_drifted_columns'] = dataset_drift.share_of_drifted_columns
            
            print(f"Dataset drift score: {dataset_drift.drift_score}")
            print(f"Number of drifted columns: {dataset_drift.number_of_drifted_columns}")
            print(f"Share of drifted columns: {dataset_drift.share_of_drifted_columns}")
            
            # Individual column drift scores
            column_drift_scores = {}
            if hasattr(dataset_drift, 'drift_by_columns'):
                for column, drift_info in dataset_drift.drift_by_columns.items():
                    column_drift_scores[column] = {
                        'drift_score': drift_info.drift_score,
                        'drift_detected': drift_info.drift_detected,
                        'p_value': getattr(drift_info, 'p_value', None)
                    }
                    print(f"Column '{column}' drift score: {drift_info.drift_score}")
            
            results['column_drift_scores'] = column_drift_scores
    
    return results
```
